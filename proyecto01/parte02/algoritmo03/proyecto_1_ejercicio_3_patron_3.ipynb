```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from pathlib import Path
```

```python
ruta_carpeta = Path(r'/content/drive/MyDrive/Colab Notebooks/redes neuronales/dataset')

archivos_excel = list(ruta_carpeta.rglob('*.xlsx'))
```

```python
lista_dfs = []

for archivo in archivos_excel:
    try:
        df = pd.read_excel(archivo)
        lista_dfs.append(df)
        print(f"Cargado: {archivo.name}")
    except Exception as e:
        print(f"Error cargando {archivo.name}: {e}")

if lista_dfs:
    datos = pd.concat(lista_dfs, ignore_index=True)
    print("\n¡Carga completa!")
else:
    print("No se encontraron archivos Excel.")
```

```python
df_final = datos
```

```python
df_final.head()
```

```python
df_final = df_final[df_final['Área'] != '9']
df_final = df_final[df_final['Sexo'] != '9']
df_final = df_final[df_final['Nivel'] != '9']
df_final = df_final[df_final['Pueblo_Per'] != '9']
df_final = df_final[df_final['Jornada_Est'] != '9']
df_final = df_final[df_final['Resultado_F'] != '9']
df_final = df_final[df_final['Repitente'] != '9']
df_final = df_final[df_final['Graduando'] != '9']
```

```python
df_final = df_final.drop(columns=[
    'Año',
    'CodEstablecimiento',
    'Departamento_F',
    'Depto_mupio',
    'Departamento',
    'Grado'
])
```

```python
df_final['feature'] = df_final['Nivel'].apply(lambda x: 1 if x == 5 else 0)
```

```python
df_final = df_final.sample(frac=1, random_state=42).reset_index(drop=True)
```

Definición de variables independientes y dependientes (a predecir); definición de conjunto de entreno y pruebas

```python
x = df_final[[
    'Pueblo_Per',
    'Plan_Est',
    'Jornada_Est',
    'Resultado_F',
    'Graduando',
    'Sector',
    'Área',
    'Sexo',
    'Repitente'
]]
y = df_final['feature']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

diseño y entreno de la red neuronal

```python
model = Sequential()
model.add(Dense(16, input_dim=9, activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test))
```

```python
loss, accuracy = model.evaluate(x_test, y_test)
print(f'loss: {loss}, Accuracy: {accuracy}')
```

predicción

```python
estudiante = np.array([[
    5,
    1,
    1,
    1,
    1,
    1,
    2,
    1,
    1
]])
res = model.predict(estudiante)
print(res)
```
